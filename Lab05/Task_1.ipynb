```python
from torch.utils.tensorboard import SummaryWriter

writer1 = SummaryWriter('runs/test1')
writer2 = SummaryWriter('runs/test2')
```

```python
import torch.nn as nn

x = torch.arange(-5, 5, 0.1).unsqueeze(1)
y = -5*x + torch.randn(len(x), 1)

MLP1 = nn.Sequential(nn.Linear(1, 5), nn.ReLU(), nn.Linear(5, 5), nn.ReLU(), nn.Linear(5, 1))
criterion = nn.MSELoss()
optimizer1 = torch.optim.SGD(MLP1.parameters(), lr=0.1)

writer1.add_graph(MLP1, x)

for epoch in range(200):
    y1 = MLP1(x)
    loss = criterion(y1, y)
    writer1.add_scalar('Loss/train', loss, epoch)
    optimizer1.zero_grad()
    loss.backward()
    optimizer1.step()
    writer1.flush()

writer1.close()
```

```python
x = torch.arange(-5, 5, 0.1).unsqueeze(1)
y = -5*x + torch.randn(len(x), 1)

MLP2 = nn.Sequential(nn.Linear(1, 10), nn.ReLU(), nn.Linear(10, 10), nn.ReLU(), nn.Linear(10, 1))
criterion = nn.MSELoss()
optimizer2 = torch.optim.SGD(MLP2.parameters(), lr=0.1)

writer2.add_graph(MLP2, x)

for epoch in range(200):
    y2 = MLP2(x)
    loss = criterion(y2, y)
    writer2.add_scalar('Loss/train', loss, epoch)
    optimizer2.zero_grad()
    loss.backward()
    optimizer2.step()
    writer2.flush()

writer2.close()
```
